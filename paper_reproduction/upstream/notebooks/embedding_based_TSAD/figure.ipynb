```python
import json
import numpy as np
import matplotlib.pyplot as plt
from dtaianomaly.data import UCRLoader
from dtaianomaly.visualization import plot_demarcated_anomalies
```

```python
def min_max(arr):
    return (arr - arr.min()) / (arr.max() - arr.min())

dataset = UCRLoader('007_UCR_Anomaly_DISTORTEDCIMIS44AirTemperature3_4000_6520_6544.txt').load()
X_test, y_test = dataset.X_test, dataset.y_test
plot_demarcated_anomalies(X_test, y_test, figsize=(20, 2))

# This is a copy of 'results\intermediate\moment_base_64\scores\474_UCR_id_172_Environment_tr_2046_1st_6520.npy',
# i.e., the anomaly scores given by MOMENT-base-64.
# We placed this file in the same directory as the notebook for convenience.
y_pred_moment = min_max(np.load('moment_base_64.npy')[-X_test.shape[0]:])

y_pred_embedding_based = min_max(np.load('embedding_based.npy'))
with open('metrics.json', 'r') as f:
    metrics = json.load(f)
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(8, 6), sharex='all')
plot_demarcated_anomalies(X_test, y_test, ax=ax1)
ax1.set_ylabel('Time series')

ax2.plot(y_pred_moment)
ax2.set_ylabel(f'MOMENT (ZS)\nVUS-PR={metrics["MOMENT (ZS)"]["VUS-PR"]:.3f}, rank={metrics["MOMENT (ZS)"]["rank"]}')

ax3.plot(y_pred_embedding_based)
ax3.set_ylabel(f'Embedding-based\nVUS-PR={metrics["Embedding-based"]["VUS-PR"]:.3f}, rank={metrics["Embedding-based"]["rank"]}')

fig.tight_layout()
```
