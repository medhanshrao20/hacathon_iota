```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from dtaianomaly.visualization import plot_demarcated_anomalies
```

```python
# Load data, labels, and anomaly scores for short and long horizons

df = pd.read_csv("../data/TSB-AD-U/558_YAHOO_id_8_WebService_tr_500_1st_1125.csv")
X = df['Data'].to_numpy()
y = df['Label'].to_numpy()

# Ensemble the scores by taking the maximum of the short and long horizon scores

def min_max(array):
    return (array - array.min()) / (array.max() - array.min())

y_pred_short_horizon = min_max(np.load(f'../results/intermediate/timesfm-1.0-200m_long_horizon_512/scores/558_YAHOO_id_8_WebService_tr_500_1st_1125/1.npy'))
y_pred_long_horizon = min_max(np.load(f'../results/intermediate/timesfm-1.0-200m_long_horizon_512/scores/558_YAHOO_id_8_WebService_tr_500_1st_1125/64.npy'))
y_pred_ensemble = np.maximum(y_pred_short_horizon, y_pred_long_horizon)
```

```python
# Visualize the original time series and the anomaly scores for the short horizon, long horizon, and ensemble

fig, (ax1, ax2, ax3) = plt.subplots(3, figsize=(8, 6), sharex='all')

plot_demarcated_anomalies(X, y, figsize=(20, 2), ax=ax1)
ax1.set_ylabel('Time series')

plot_demarcated_anomalies(y_pred_short_horizon, y, ax=ax2)
ax2.plot(y_pred_short_horizon, color='tab:blue',label='$h=1$') # This line is redundant but it ensures the legend is correct
ax2.plot(y_pred_long_horizon, color='tab:orange', label='$h=64$')
ax2.legend()
ax2.set_ylabel(f'Anomaly score\nsingle horizon')

plot_demarcated_anomalies(y_pred_ensemble, y, ax=ax3)
ax3.set_ylabel(f'Anomaly score\nmultiple horizons')

fig.tight_layout()
```
