```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Boxplot of VUS-PR scores for all variants

scores = []

variants_tsfm = ['bolt_small_64', 'bolt_small_256', 'bolt_small_512', 'bolt_base_64', 'bolt_base_256', 'bolt_base_512', 'TimeMoE-50M_64', 'TimeMoE-50M_256', 'TimeMoE-50M_512', 'TimeMoE-200M_64', 'TimeMoE-200M_256', 'TimeMoE-200M_512', 'timesfm-1.0-200m_64', 'timesfm-1.0-200m_256', 'timesfm-1.0-200m_512', 'timesfm-2.0-500m_64', 'timesfm-2.0-500m_256', 'timesfm-2.0-500m_512']
variants_baseline = ['last_1', 'last_3', 'centered_5']
variants = variants_tsfm + variants_baseline

for variant in variants_tsfm:
    score = pd.read_csv(f'../results/tsfms/{variant}.csv')['VUS-PR'].to_numpy()
    scores.append(score)

for variant in variants_baseline:
    score = pd.read_csv(f'../results/baselines/{variant}.csv')['VUS-PR'].to_numpy()
    scores.append(score)

plt.figure(figsize=(6, 8))
plt.rcParams.update({'font.size': 15})
plt.boxplot(scores[::-1], showmeans=True, meanline=True, vert=False, tick_labels=variants[::-1])
plt.xlabel("VUS-PR Score")
plt.grid(False)
plt.show()
```

```python
# Plot heatmap of the correlation matrix between variants

scores = np.array(scores).astype(np.float32)
corr_matrix = np.corrcoef(scores)

fig, ax = plt.subplots(figsize=(20, 16))
cax = ax.imshow(corr_matrix, cmap='coolwarm', vmin=-1, vmax=1)
cbar = fig.colorbar(cax, ax=ax)

# Set ticks and labels
num_methods = scores.shape[0]
ticks = np.arange(num_methods)
ax.set_xticks(ticks)
ax.set_yticks(ticks)
ax.set_xticklabels(variants, rotation=45, ha='right')
ax.set_yticklabels(variants)

# Add annotations (correlation values)
for i in range(num_methods):
    for j in range(num_methods):
        ax.text(j, i, f"{corr_matrix[i, j]:.2f}", ha='center', va='center', color='black', fontsize=15)

plt.tight_layout()
plt.show()
```
